```python
import pandas as pd
import numpy as np

from utils import Display
```

```python
city = '北京'
```

```python
file_path = '/mnt/data/raydata/政策-经济-建设/各级别产业园区信息（2025.7更新）/国家级、省级、区县级产业园区数据.xlsx'
park = pd.read_excel(file_path)
```

```python
Display(park)
```

```python
housing_path = f'/mnt/data/raydata/房产/Housing_Sale_Listing_Rent/listing_p/{city}.parquet'
housing = pd.read_parquet(housing_path)
Display(housing)
```

```python
from utils import data_cleaning
from sklearn.neighbors import NearestNeighbors

sub_park, sub_housing = data_cleaning(park, housing)

park_coords_rad = np.radians(sub_park[['x', 'y']])
house_coords_rad = np.radians(sub_housing[['x', 'y']])

model = NearestNeighbors(n_neighbors=1, algorithm='ball_tree', metric='haversine')
model.fit(park_coords_rad)

distances_rad, indices = model.kneighbors(house_coords_rad)

sub_housing['nearest_park_dist_km'] = distances_rad.flatten() * 6371

sub_housing['nearest_park_index'] = sub_park.index[indices.flatten()]

sub_housing['nearest_park_industry'] = sub_park.loc[sub_housing['nearest_park_index'], '产业'].values
print(sub_housing[['x', 'y', 'nearest_park_index', 'nearest_park_dist_km']].head())
```

```python
from utils import process_num

sub_housing['unit_price'] = sub_housing['unit_price'].apply(process_num)
sub_housing['log_price'] = np.log(sub_housing['price'])
sub_housing['log_unit_price'] = np.log(sub_housing['unit_price'])
```

```python
# plot the distribution of price , log_price , unit price and unit_log_price
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(12, 5))
sns.histplot(sub_housing['price'], bins=50, ax=axes[0,0], kde=True)
axes[0,0].set_title('Price Distribution') 
sns.histplot(sub_housing['unit_price'], bins=50, ax=axes[0,1], kde=True)
axes[0,1].set_title('Unit Price Distribution')
sns.histplot(sub_housing['log_price'], bins=50, ax=axes[1,0], kde=True)
axes[1,0].set_title('Log Price Distribution')
sns.histplot(sub_housing['log_unit_price'], bins=50, ax=axes[1,1], kde=True)
axes[1,1].set_title('Log Unit Price Distribution')
plt.tight_layout()
plt.show()
```

```python
sub_housing['area'] = sub_housing['price'] * 10000 / sub_housing['unit_price']
```

```python
print(sub_housing['area'].describe())
```

```python
# plot the distribution of area
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.histplot(sub_housing['area'], bins=50, kde=True)
plt.title('Area Distribution')
plt.xlabel('Area (sqm)')
plt.ylabel('Frequency')
plt.show()
```

```python
from pathlib import Path
output_path = Path(f"processed_data/{city}")
output_path.mkdir(parents=True, exist_ok=True)

sub_housing.to_csv(output_path / '01_housing.csv', index=False)
sub_park.to_csv(output_path / '01_park.csv', index=False)
```
